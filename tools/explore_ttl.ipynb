```python
from rdflib import Graph

g = Graph()
nodes = {}

def parse_rdf(ttl_file):
    global g
    with open(ttl_file, "r") as ttl:
        lines = ttl.read()
    graph = g.parse(data=lines, format="turtle")
    return graph

g = parse_rdf("VOCAB_SP223_enumeration-v1.0.ttl")
```

```python
for s,p,o in g:
    if 'Electricity-AC' in o:
        title = s.split('#')[1]
    elif 'Electricity-DC' in o:
        title = s.split('#')[1]
    details = "3 Phases" if '3Ph' in title else "1 phase"
    class_name = f"Electricity_{title.split('AC')[1].replace('-','_')}"
    medium = f"Electricity.{title}"
```
