# Tutorial 009 - Layering other models

One of the features of the Semantic Web is the idea that other ontologies can be used at the same time in the same graph.  One of those that the Strike Team has been testing is Real Estate Core (REC) to represent finer grained representations of physical spaces.  The 223 model has only one abstract concept called a `PhysicalSpace` and REC has an entire taxonomy of them.

```python
from bob.core import bind_model_namespace, dump
from bob.space.physical import Building, Floor, Office

_namespace = bind_model_namespace("ex", "http://example/")
```
