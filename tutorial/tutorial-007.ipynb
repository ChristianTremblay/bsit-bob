# Tutorial 007 - Building Systems

This tutorial creates a type of `System` that contains two devices and has system connection points mapped to the device inputs and outputs.

```python
from bob.core import Device, System, bind_model_namespace, dump
from bob.connections.air import AirInletConnectionPoint, AirOutletConnectionPoint
from bob.connections.air import AirInletSystemConnectionPoint, AirOutletSystemConnectionPoint

_namespace = bind_model_namespace("ex", "http://example/")
```

First define the types of devices like before:

```python
class OutputDevice(Device):
    aocp: AirOutletConnectionPoint

class InputDevice(Device):
    aicp: AirInletConnectionPoint
```

Now make a new kind of system that contains one of each of these devices:

```python
class SomeSystem(System):
    ai: InputDevice
    ao: OutputDevice
    aicp: AirInletSystemConnectionPoint
    aocp: AirOutletSystemConnectionPoint

    def __init__(self, *, system_label: str) -> None:
        super().__init__(label=system_label)
        
        self.ai = InputDevice(label=system_label + ".ai")
        self > self.ai
        
        self.ao = OutputDevice(label=system_label + ".ao")
        self > self.ao
        
        # map the system connection points to the devices' connection points
        self.aicp.mapsTo = self.ai.aicp
        self.aocp.mapsTo = self.ao.aocp
```

Making a pair of these systems is easy, and instances of all of their constitutent devices are also created:

```python
sys1 = SomeSystem(system_label="sys1")
print(f"{sys1.ao=}")

sys2 = SomeSystem(system_label="sys2")
print(f"{sys2.ai=}")
```

And now the two instances can be connected together:

```python
sys1 >> sys2
```
