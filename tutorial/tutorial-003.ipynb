# Tutorial 003 - Custom Device

In this tutorial we are going to define our own type of device, make and instance of it, and
then dump out the contents of the data graph and the schema graph.

> The _data graph_ is a all of the instances of the "things" in the model, the _schema graph_ is a description of what those things are.  If 223 was using the OWL schema language then the schema graph would be an instance of an OWL Ontology.

```python
from bob.core import Device, bind_model_namespace, dump, data_graph, schema_graph
```

Having blank node identifiers in graphs is necessary, but it's quite annoying when analyzing and validating a graph because the URIs show up in reports.  To make this easier to read and understand, first create a namespace for our examples and Bob will use it when creating blank nodes:

```python
_namespace = bind_model_namespace("ex", "http://example/")
```

Now we are going to create a subclass of `Device` and an instance, note that the IRI is easier to read:

```python
class MyDevice(Device):
    pass

d = MyDevice(label="d")
print(f"{d=}")
```

Now we can investigate the graph of the new thing we created:

```python
dump(data_graph)
```

```python
dump(schema_graph)
```

The astute RDF entheusist will note that the triple `ex:00001 a s223:Device` is in some sense redundant because saying that `ex00001 a ex:Device` and that `ex:MyDevice a rdfs:subClassOf s223:Device` is enough with tools that have RDFS reasoning built in, in fact `ex:MyDevice a rdfs:Class` is also redundant since `ex:MyDevice` is a subclass of something implies that it is an `rdfs:Class`.

> There are many places where Bob generates more RDFS statements than absolutely necessary because many tools do _not_ have RDFS inferencing built in amd it can be computationally expensive to run an RDFS reasoner.  There are additional inference rules built into the 223 model that provide additional useful statements that are necessary for resolving competency questions and common queries without the computational overhead.
