# Tutorial 008 - External References

The 223 standard describes an `ExternalReference` as a generic class that is intended to be subclassed for reading (observable properties) and writing (actuatable properties) using other protocols.  One of those is BACnet and support for BACnet external references is built into Bob.

```python
from bob.core import bind_model_namespace, dump
from bob.externalreference.bacnet import BACnetReference

_namespace = bind_model_namespace("ex", "http://example/")
```

While it is possible to fill out the contents of a `bacnet:DeviceObjectPropertyReference` with the individual fields, it is simpler to use the pattern described in **Annex XX**:

```python
ref = BACnetReference("bacnet://12345/analog-value,1")
dump()
```
