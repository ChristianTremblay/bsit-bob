# Tutorial 006 - Connecting Chains of Equipment

This tutorial builds on the previous one, but showing other equipment in a sequence.

```python
from bob.core import Equipment, Property, bind_model_namespace, dump
from bob.connections.air import AirInletConnectionPoint, AirOutletConnectionPoint
from bob.equipments.hvac.filter import Filter

_namespace = bind_model_namespace("ex", "http://example/")
```

In this case we'll use a `Filter` class from the "proposed" namespace, a collection of different kinds of things that have not been completely vetted by the committee but some members have found useful in building models.

```python
class OutputEquipment(Equipment):
    aocp: AirOutletConnectionPoint

class InputEquipment(Equipment):
    aicp: AirInletConnectionPoint

out_equipment = OutputEquipment(label="out_equipment")
in_equipment = InputEquipment(label="in_equipment")
```

Now take the output from the `out_equipment` and run it through the filter before sending it into the `in_equipment` showing how Equipments can be chained together:

```python
out_equipment >> Filter(label="f") >> in_equipment
```

And to make it even easier, an `OutputEquipment` can send it's output to more than one `InputEquipment` through the same connection, like the output of an **AHU** going to two separate **VAV** terminal units:

```python
OutputEquipment(label="AHU") >> [ InputEquipment(label="VAV1"), InputEquipment(label="VAV2") ]
```

Dump the results and give the TTL file to your favorite semantic web tool to visualize the results:

```python
dump()
```
