# Tutorial 002 - Equipment Instance

In this tutorial we are going to make an instance of a `s223:Equipment` and give it a `s223:Property`, and dump out the data graph.

```python
from bob.core import Equipment, Property, dump
```

```python
d = Equipment(label="d")
```

Being an instance of a `Equipment` is more interesting that just being a node, so there is a statement to the effect that shows up in the graph.  The `label` parameter is so useful when browsing around the model using other tools that is a required parameter of a `Equipment` (and all its subclasses).  It is expressed in the graph as an `rdfs:label` predicate and using concepts from other models or ontologies is central to the Semantic Web.

The blank node IRI assigned to `d` shows up as a blank node in the resulting turtle file:

```python
dump()
```

Now create a property and give it a value.  The parameter value can be anything that can be turned into an RDF literal:

```python
p = Property(75.3)
dump()
```

Now we can add the property to the Equipment, and there is an API to do that:

```python
d.add_property(p)
dump()
```
