# Tutorial 001 - Nodes in a graph

**Bob** is a Python package for quickly creating instances of **Semantic Interoperability Working Group** models which are **RDF** documents that describe Equipments, systems as collections of Equipments, and the properties of those things.  As the SI-WG model evolves it provides a simple way of rebuilding model instances to verify that earlier assumptions about how things are linked together and removes some of the otherwise tedious work of working with the model.

The foundation of Bob is a **Node** and instances of a Node are simultaneously defined in a _data graph_.  Bob provides easy access to `dump()` the contents of the data graph, and if there are more interesting things to do with the graph it is just an instance of a `Graph` as defined in **rdflib**.

> The concept of a Node is not defined in 223, it is used as a base class for all "things" in Bob which are described in RDF as the subjects or objects of a `(subject, predicate, object)` statement, a.k.a triple.

```python
from bob.core import Node, data_graph
```

Now create a `Node` and note that the data graph is empty (the length of a graph in `rdflib` is the number of statements):

```python
x = Node()
print(len(data_graph))
```

> The fact that `x` is a node isn't very interesting, so some statement to that like having a triple `_:x a bob:Node` is not included.  There are also other abstract concepts in 223 that are also subclasses of Node in Bob that will not generate statements.

The created node has an IRI that is a private attribute of the node, so when it does show up in the graph you'll known which one it is, very important for debugging!

```python
print(f"{x=}")
print(f"{x._node_iri=}")
```

Unless a node has been given a specific IRI for some reason, it will be a blank node.  Here is another node instance that refers to some "resource" someplace out in the network:

```python
from rdflib import URIRef

y = Node(_node_iri=URIRef("mailto:peter.parker@dailybugle.com"))
print(f"{y=}")
```
