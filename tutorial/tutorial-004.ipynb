# Tutorial 004 - Custom Property

In this tutorial we are going to define our own type of device with a special property, make and instance of it, and
then dump out the contents of the data graph and the schema graph.

First, go through the same steps as before importing and creating a namespace:

```python
from bob.core import Device, Property, bind_model_namespace, dump, data_graph, schema_graph
_namespace = bind_model_namespace("ex", "http://example/")
```

Now we are going to create a subclass of `Device` and an instance, note that the IRI is easier to read:

```python
class MyDevice(Device):
    age: Property

d = MyDevice(label="d")
print(f"{d=}")
print(f"{d.age=}")
```

Now we can investigate the graph of the new thing we created:

```python
dump(data_graph)
```

```python
dump(schema_graph)
```

Notice that the device doesn't have an "age" yet, but the schema says that `ex:age` is a subproperty of `s223:Property`.  So give it an age like any Python programmer would expect to be able to do:

```python
d.age = 75.3
dump()
```

Bob did a whole bunch of things, like seeing that the programmer provided a value that needed to be converted to an instance of a `s223:Property`, then making a statement about the device having a property with that value, _and_ making a statement in our example model using `ex:age`.  Now if we look at the `age` attribute of the node we get specialized subclass of `Node`:

```python
print(f"{d.age=}")
```

Bob is used to build models that are expected to be relatively static, not necessarily containing "telemetry data."  There are some exceptions, but once an attribute is given a value it is not expected to be given another value:

```python
d.age = 12.5
```
